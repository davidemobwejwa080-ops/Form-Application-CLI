```python


def get_name():
    name = input("Enter your full name: ")
    return name


def get_age():
    age = int(input("Enter your age: "))
    return age


def get_gender():
    gender = input("Enter your gender: ")
    return gender


def get_work_status():
    print("\nSelect your work status:")
    print("1. Employed")
    print("2. Unemployed")
    print("3. Student")
    print("4. Self-employed")

    choice = input("Enter your choice (1-4): ")

    statuses = {
        "1": "Employed",
        "2": "Unemployed",
        "3": "Student",
        "4": "Self-employed"
    }

    return statuses.get(choice, "Not specified")


def get_email():
    email = input("Enter your email address: ")
    return email


def generate_card(name, age, gender, work_status, email):
    print("\n" + "=" * 40)
    print("          FORM APPLICATION")
    print("=" * 40)

    print(f"Full Name   : {name}")
    print(f"Age         : {age}")
    print(f"Gender      : {gender}")
    print(f"Work Status : {work_status}")
    print(f"Email       : {email}")

    print("=" * 40)
    print("     APPLICATION COMPLETED!")
    print("=" * 40)


def main():
    print("WELCOME TO FORM APPLICATION")
    print("Please provide your information.\n")

    name = get_name()
    age = get_age()
    gender = get_gender()
    work_status = get_work_status()
    email = get_email()

    generate_card(name, age, gender, work_status, email)


main()
```
